```python
def count_missing(x):
    return x.isnull().sum()

df_orders.groupby('order_status')[
    ['order_approved_at',
     'order_delivered_carrier_date',
     'order_delivered_customer_date']
].apply(count_missing)
```

```python
df_orders[
    (df_orders['order_status'] == 'delivered') &
    (
        df_orders['order_approved_at'].isnull() |
        df_orders['order_delivered_carrier_date'].isnull() |
        df_orders['order_delivered_customer_date'].isnull()
    )
]
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

df_orders[date_columns] = df_orders[date_columns].apply(
    pd.to_datetime
)
```

```python
df_orders.info()
```

```python
df_orders[
    (df_orders['order_approved_at'] < df_orders['order_purchase_timestamp']) |
    (df_orders['order_delivered_carrier_date'] < df_orders['order_approved_at']) |
    (df_orders['order_delivered_customer_date'] < df_orders['order_delivered_carrier_date'])
].shape
```

```python
print("Approved before Purchase:",
      (df_orders['order_approved_at'] < df_orders['order_purchase_timestamp']).sum())

print("Carrier before Approved:",
      (df_orders['order_delivered_carrier_date'] < df_orders['order_approved_at']).sum())

print("Customer Delivery before Carrier:",
      (df_orders['order_delivered_customer_date'] < df_orders['order_delivered_carrier_date']).sum())
```

```python
df_orders[
    df_orders['order_delivered_carrier_date'] <
    df_orders['order_approved_at']
][[
    'order_id',
    'order_status',
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date'
]].head(10)
```

```python
df_orders[
    df_orders['order_delivered_customer_date'] <
    df_orders['order_delivered_carrier_date']
][[
    'order_id',
    'order_status',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date'
]].head(10)
```

### Orders Table — Key Findings

- The orders table contains **99,441 records and 8 columns**.
- The majority of orders are in the **delivered** status (96,478), while the remaining orders are distributed across shipped, canceled, unavailable, invoiced, processing, created, and approved statuses.
- Missing values are mainly present in approval and delivery timestamps and are largely associated with orders that were not completed or delivered.
- A small number of delivered orders also contain missing timestamps, which were retained rather than removed.
- All five timestamp columns were converted from `object` to `datetime` format for accurate time-based analysis.
- Chronological inconsistencies were detected in some timestamps, particularly carrier and customer delivery dates. These records were retained and documented rather than modifying dates without reliable evidence.

**Conclusion:**

The orders table is suitable for further analysis after datetime conversion and data-quality validation. The table can be used to analyze order trends, delivery performance, order status, and operational efficiency.
