```python

df_customers = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_customers_dataset.csv")
df_sellers = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_sellers_dataset.csv")
df_reviews = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_order_reviews_dataset.csv")
df_items = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_order_items_dataset.csv")
df_products = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_products_dataset.csv")
df_geolocation = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_geolocation_dataset.csv")
df_category = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/product_category_name_translation.csv")
df_orders = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_orders_dataset.csv")
df_payments = pd.read_csv("/kaggle/input/datasets/olistbr/brazilian-ecommerce/olist_order_payments_dataset.csv")
```

```python
datasets = {
    "Customers": df_customers,
    "Sellers": df_sellers,
    "Reviews": df_reviews,
    "Order Items": df_items,
    "Products": df_products,
    "Geolocation": df_geolocation,
    "Category Translation": df_category,
    "Orders": df_orders,
    "Payments": df_payments
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```
